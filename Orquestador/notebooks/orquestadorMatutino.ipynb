```python
from IPython import get_ipython
from datetime import datetime

def run_notebook(path):
    print(f"\n=== Running {path} ===")
    start = datetime.now()

    get_ipython().run_line_magic("run", f'"{path}"')

    elapsed = datetime.now() - start
    print(f"=== Finished {path} | ⏱ {elapsed} ===\n")
    return elapsed

notebooks = [
    "../CasosDeUso/CasosErick/Vistas KPIs.ipynb",
    "../CasosDeUso/CasosErick/Dashboard_V2.ipynb",
    "../CasosDeUso/CasosErick/DEV_Dashboard_V2.ipynb",
    "../CasosDeUso/CasosErick/Estatus de pedidos.ipynb",
    "../CasosDeUso/Dashboards/PanelActividad.ipynb",
    "../CasosDeUso/Dashboards/kpis_celulaCredito.ipynb"
]

celda_start = datetime.now()

for nb in notebooks:
    try:
        print(f"Corriendo {nb}")
        elapsed = run_notebook(nb)
        print(f"Finalizando {nb} | ⏱ {elapsed}\n\n")
    except Exception as e:
        print(f"❌ {nb} failed: {e}")
        break

total = datetime.now() - celda_start
print(f"\n⏱ Tiempo total de ejecución: {total}")
```
